## 🌟 What is LangChain?

**LangChain** is an open-source framework that helps developers build **Large Language Model (LLM)** applications.
It simplifies the process of connecting AI models with **external tools**, **data sources**, and **real-world
tasks**.

Think of it as a **"bridge"** between your AI model and the world — allowing it to:
- Read from and write to files
- Search the internet
- Query databases
- Use APIs
- Follow instructions from users
- Chain multiple AI operations together

---

## 🧠 Why Use LangChain?

| ✅ Benefit | What It Means |
|-----------|----------------|
| **Modularity** | Reuse components like chains, prompts, and agents |
| **Interoperability** | Connect with many models (OpenAI, HuggingFace, etc.) |
| **Simplicity** | Less boilerplate code to write from scratch |
| **Extensibility** | Build complex workflows step-by-step |

## 🏗️ How LangChain Works

```
┌──────────────────┐
│   User Input     │
└────────┬─────────┘
         │
         ▼
┌──────────────────┐
│  LLM/Model       │
└────────┬─────────┘
         │
         ▼
┌──────────────────┐
│  Chain/Agent     │
└────────┬─────────┘
         │
         ▼
┌──────────────────┐
│  Data Source     │
└──────────────────┘
```

1. **User** provides input
2. **LLM** generates text
3. **Chain** orchestrates multiple steps
4. **Data Source** provides context (files, API, web)

## 🔍 Component Breakdown

| Parameter | Value | Purpose |
|-----------|-------|---------|
| **model** | `"llama3.2:3b"` | Specifies which Llama 3.2 model (3 billion parameters) to use |
| **temperature** | `0.1` | Controls randomness; low value = more deterministic responses |
| **base_url** | `"http://127.0.0.1:11435"` | Ollama local server address (localhost port 11435) |

---

## 🧠 Understanding Each Parameter

### 1️⃣ `model="llama3.2:3b"`

This selects **Llama 3.2 3B**, a smaller but capable language model:
- **3B** = 3 billion parameters (lighter than 7B or 8B models)
- Faster inference and lower memory usage
- Still powerful for most common tasks
- **Note**: You must run `ollama pull llama3.2:3b` first

### 2️⃣ `temperature=0.1`

This sets a **very low temperature** for the LLM:
```
Higher temperature (0.9-1.0) → More creative, varied responses
Lower temperature (0.0-0.2) → More consistent, focused responses
```

For **production tools**, you want:
- ✅ Predictable outputs
- ✅ Consistent tool behavior
- ✅ Less hallucination
- ✅ Better reliability

### 3️⃣ `base_url="http://127.0.0.1:11435"`

This connects to your **local Ollama server**:
```
127.0.0.1 = localhost (your own computer)
11435 = default Ollama API port
```

You can start Ollama with:
```bash
ollama serve
```
Or simply:
```bash
ollama run llama3.2:3b
```

# Code Explanation: Domain-Specific LLM Tool Definitions

This code snippet defines a set of specialized **tools** for a Large Language Model (LLM) application. It utilizes a common pattern in AI agent
frameworks (such as **LangChain**) to create "experts" by assigning specific personas to a single underlying language model instance.

---

## 🏗️ Architecture Overview

The code registers three callable functions as **tools**. In the context of LLM agents, a "tool" is a function the agent can choose to call when a
user asks a specific type of question.

| Component | Description |
| :--- | :--- |
| **`expert_llm`** | A pre-configured instance of an LLM that the tools invoke. |
| **`@tool`** | A decorator used to register the function as a callable action for the agent. |
| **`QueryInput`** | A Pydantic (or similar) schema defining the input type (likely `str`). |
| **Prompts** | Dynamic strings that assign a specific "persona" (role) to the LLM for each task. |

---

## 🔍 Code Breakdown

### 1. The `@tool` Decorator

```python
@tool(args_schema=QueryInput)
```

*   **Purpose:** Marks the function as a Tool available to the orchestration framework.
*   **`args_schema`**: Defines the structure of the input arguments. Here, it expects a simple query string.
*   **Effect:** When a user asks a question, the agent looks at available tools to see which one is most appropriate.

### 2. The Function Signatures

Each function follows this structure:
*   **Input:** `query: str` — A text prompt from the user or agent.
*   **Logic:** Invokes `expert_llm` with a system prompt injected via an f-string.
*   **Output:** `str` — Returns the raw text content of the LLM's response.

### 3. Specific Experts

Each tool uses the same LLM instance (`expert_llm`) but changes the **System Prompt** to specialize the output.

| Tool Name | Persona Role | System Prompt Strategy |
| :--- | :--- | :--- |
| **`math_expert`** | **Mathematician** | Instructs the LLM to solve problems **step-by-step**. |
| **`history_expert`** | **Historian** | Instructs the LLM to provide **factual and accurate** answers. |
| **`coding_expert`** | **Software Engineer** | Instructs the LLM to act as a **senior developer**. |

---

## 🚀 Execution Flow

When a user interacts with the application, the logic flows as follows:

1.  **Query Ingestion:** The user asks a question (e.g., *"What is the capital of France?"* or *"Solve 2x + 5 = 15"*).
2.  **Tool Selection:** The AI Agent evaluates the query against the available tools.
    *   *Mathematical query* → Triggers `math_expert`.
    *   *Historical query* → Triggers `history_expert`.
3.  **Role Assignment:** The agent constructs the prompt:
    > "You are an expert mathematician. Solve step-by-step:\n{user_query}"
4.  **Invocation:** The agent sends the prompt to `expert_llm`.
5.  **Response Handling:** The raw text (`response.content`) is returned and sent back to the user.

---

## 💡 Key Takeaways

1.  **Single Model, Multiple Personalities:** Instead of loading three different expensive LLM models, this pattern routes queries to one model while
changing the **system instruction** (the persona). This is cost-effective and efficient.
2.  **Structured Inputs:** By using `args_schema`, the agent can parse user inputs into valid argument structures, preventing hallucinated or broken
inputs.
3.  **Separation of Concerns:** Each tool handles a distinct domain, reducing the likelihood of the model making errors (e.g., the model isn't
expected to know deep math history, just math).
4.  **Extensibility:** To add a new capability (e.g., `science_expert`), you would simply copy one of the function blocks, rename it, and change the
system prompt instruction.

---

````python
# ==============================
# Tool Definitions
# ==============================
@tool(args_schema=QueryInput)
def math_expert(query: str) -> str:
    """Solve mathematical problems."""
    response = expert_llm.invoke(
        f"You are an expert mathematician. Solve step-by-step:\n{query}"
    )
    return response.content


@tool(args_schema=QueryInput)
def history_expert(query: str) -> str:
    """Answer historical questions with accurate facts."""
    response = expert_llm.invoke(
        f"You are a historian. Provide a clear and factual answer:\n{query}"
    )
    return response.content


@tool(args_schema=QueryInput)
def coding_expert(query: str) -> str:
    """Answer programming and technical questions."""
    response = expert_llm.invoke(
        f"You are a senior software engineer. Help with this:\n{query}"
    )
    return response.content
````

````python
# ==============================
# Main Application
# ==============================
def main():
    print("\n" + "=" * 70)
    print("LANGCHAIN LOCAL OLLAMA AGENT (LLM-POWERED TOOLS)")
    print("=" * 70 + "\n")

    try:
        # Orchestrator LLM
        print("Initializing Orchestrator...")
        orchestrator_llm = ChatOllama(
            model="llama3.2:3b",
            temperature=0.5,
            base_url="http://127.0.0.1:11435"
        )
        print("✓ Orchestrator ready\n")

        # Bind tools
        tools = [math_expert, history_expert, coding_expert]
        llm_with_tools = orchestrator_llm.bind_tools(tools)

        print("=" * 70)
        print("AGENT READY - Tools:")
        print("• math_expert")
        print("• history_expert")
        print("• coding_expert")
        print("\nType 'exit' to quit")
        print("=" * 70 + "\n")

        # Tool mapping (clean execution)
        tool_map = {
            "math_expert": math_expert,
            "history_expert": history_expert,
            "coding_expert": coding_expert,
        }

        # Main loop
        def userInputs(query :str):
        
            try:
                user_input = query

                print("\n[Processing...]")

                # Add system + user message
                messages = [
                    SystemMessage(content="""
You are an intelligent orchestrator.

Routing rules:
- Use math_expert → math problems
- Use history_expert → history questions
- Use coding_expert → programming questions

IMPORTANT:
- Always pass arguments as: {"query": "<user question>"}
- DO NOT generate structured objects like {'type': ...}
- Always follow tool schema strictly
"""),
                    HumanMessage(content=user_input)
                ]

                # First LLM call
                response = llm_with_tools.invoke(messages)

                # Tool loop
                while hasattr(response, "tool_calls") and response.tool_calls:
                    print("\n[Tool Execution]")

                    tool_messages = []

                    for tool_call in response.tool_calls:
                        tool_name = tool_call["name"]
                        tool_args = tool_call["args"]

                        # Safe extraction
                        if isinstance(tool_args, dict) and "query" in tool_args:
                            query_arg = tool_args["query"]
                        else:
                            query_arg = str(tool_args)

                        print(f"→ Using {tool_name}: {query_arg}")

                        tool_fn = tool_map.get(tool_name)

                        if tool_fn:
                            result = tool_fn.invoke({"query": query_arg})
                        else:
                            result = "Unknown tool"

                        print("✓ Tool executed")

                        tool_messages.append(
                            ToolMessage(
                                content=result,
                                tool_call_id=tool_call["id"]
                            )
                        )

                    # Append history
                    messages.append(response)
                    messages.extend(tool_messages)

                    # Re-call LLM
                    response = llm_with_tools.invoke(messages)

                # Final output
                print("\n[Final Answer]")
                print(response.content)
                print("\n" + "-" * 70)

            except Exception as error:
                print(f"\nError: {error}\n")

        # Run the agent with user input
        userInputs("What is (64*4/2+10)?")
        userInputs("Who was the first president of the United States?")
        userInputs("Write a Python code to reverse a string.")

        
    except Exception as init_error:
        print(f"\nInitialization Error: {init_error}")
        print("\nMake sure:")
        print("1. Ollama is running")
        print("2. Model exists: ollama pull llama3.2:3b")
        print("3. Correct port: 11435")
````

````python
if __name__ == "__main__":
    main()
````
